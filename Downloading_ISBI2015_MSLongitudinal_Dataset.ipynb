# ISBI 2015 Multiple Sclerosis MRI Dataset

This notebook downloads and prepares the **official ISBI 2015 Longitudinal Multiple Sclerosis Lesion Segmentation Challenge dataset** for research.

### Dataset

- **5 training + 14 test subjects**
- **T1, T2, PD, and FLAIR MRI**
- Manual lesion segmentations for the training data

### This Notebook

- Connects Google Colab to Google Drive
- Downloads the official dataset
- Verifies and extracts the files
- Stores the dataset permanently in Google Drive

### Official Source

https://iacl.ece.jhu.edu/index.php/MSChallenge/data

## 1. Mount your Google Drive

```python
from google.colab import drive

drive.mount('/content/drive')
```

## 2. Create a permanent folder in Drive

```python
import os

DATA_DIR = "/content/drive/MyDrive/ISBI_2015_MS"

os.makedirs(DATA_DIR, exist_ok=True)

print(DATA_DIR)
```

```python
!wget -c \
"https://iacl.ece.jhu.edu/~aaron/data/training_final_v4.zip" \
-O "/content/drive/MyDrive/ISBI_2015_MS/training_final_v4.zip"
```
